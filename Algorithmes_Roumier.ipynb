```python
def somme_n():
    "Méthode avec boucle Pour"
    n=int(input("Veuillez saisir un entier n :"))
    somme=0
    for i in range (1,n+1):
        somme=somme+i
    return somme

def somme_while(n):
    "Méthode avec la boucle Tant que"
    somme=0
    i=1
    while i<=n :
        somme=somme+i
        i=i+1
    return somme

somme_while(5)
```

```python
def moyenne_n(n) :
    """Version itérative"""
    if n==0 :
        print ('Erreur : n doit etre superieur à 0')
        return None
    somme=0
    for i in range (1, n+1):
        somme=somme+i
    moyenne=somme/n
    print ("Somme :", somme)
    return moyenne

def somme_n_rec(n) :
    if n==0 :
        return 0
    return (somme_n_rec(n-1)+n)

def moyenne_n_rec() :
    """Version récursive"""
    n = int(input ("Saisir n"))
    if n==0 :
        print ("Erreur, n doit etre superieur à 0")
        return None
    return (somme_n_rec(n))/n

moyenne_n_rec()
```

```python
def Fact_n(n):
    "Méthode itérative"
    if n==0:
        return (1)
    fact=1
    for i in range (1,n+1):
        fact=fact*i
    print ("Le factoriel de ", n, " par la méthode itérative est :", fact)
    return fact

Fact_n(5)

def Fact_rec(n):
    "Méthode récursive"
    if n==0 :
        return (1)
    return (Fact_rec(n-1)*n)
```

```python
def PlusBase2(x,y):
    z=x+y
    Reste = z-(z//2)*2
    return Reste

def Plus_Base2(x,y):
    "Version en testant si la somme est paire"
    z=x+y
    return (z%2)

def MultBase2(x,y):
    z=x*y
    Reste=z-(z//2)*2
    return Reste

def Mult_Base2(x,y):
    "Version en testant si le produit est pair"
    z=x*y
    return (z%2)

Mult_Base2(5,7)
```

```python
def Fib(n) :
    "Méthode récursive"
    if n==0 :
        return 0
    if n==1 :
        return 1
    return (Fib(n-1)+Fib(n-2))

def Fib_it(n):
    "Méthode itérative"
    if n==0:
        return 0
    if n==1 :
        return 1
    fib2=0 #Valeur du terme n-2 de la suite de Fibonacci
    fib1=1 #Valeur du terme n-1 de la suite de Fibonacci
    for i in range (2, n+1):
        fib=fib1+fib2 #Valeur du terme n de la suite de Fibonacci
        fib2=fib1 #Le terme n-2 prend maintenant la valeur du terme i-1
        fib1=fib #Le terme n-1 prend maintenant la valeur du terme i avant que i soit incrémenté de 1
    return fib

Fib(11)
    
```
